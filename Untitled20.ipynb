```python
# %% [markdown]
# # 11주차: DAG와 복합표본 회귀 (4팀: 현재흡연 → 복부비만)
# 자료: 국민건강영양조사 제9기 3차년도(2024). 1–3번 셀은 **10주차 노트북(`10_table1.ipynb`)의 셀 1–3을 그대로** 가져왔다(불러오기, 변수 만들기, `in_analysis`).

# %% [markdown]
# ## 1. 자료 불러오기 (10주차 셀 1)

# %% 셀 1
# pyreadstat 라이브러리가 없다면 자동으로 설치합니다
!pip install pyreadstat -q

import pandas as pd
import numpy as np
import pyreadstat
from google.colab import drive

# Google Drive 연동
drive.mount('/content/drive')

# 파일 경로 (파일을 우클릭해서 복사한 경로를 여기에 붙여넣으세요)
file_path = '/content/drive/MyDrive/hdac_data/hn24.sav'

# 원시자료 불러오기 (메타데이터와 데이터프레임 분리)
raw_df, meta = pyreadstat.read_sav(file_path)
print("원시자료 행/열 크기:", raw_df.shape)

# %% [markdown]
# ## 2. 노출·결과·공변량 만들기 (10주차 셀 2)

# %% 셀 2
# [셀 2] 변수 추출 및 재코딩 (9주차 프로토콜 반영)
vars_needed = ['year', 'kstrata', 'psu', 'wt_itvex', 'age', 'sex', 'BS3_1', 'BS1_1', 'HE_wc', 'incm', 'edu', 'BD1_11', 'pa_aerobic']
df = raw_df[vars_needed].copy()

# 1. 노출 정의: BS3_1 (1, 2 -> 현재흡연 1 / 3, 8 -> 비흡연 0 / 9 및 결측 -> 제외)
def recode_smoking(row):
    bs3 = row['BS3_1']
    if bs3 in [1, 2]:
        return 1  # 현재흡연군
    elif bs3 in [3, 8]:
        return 0  # 비흡연군 (과거흡연 + 평생 비흡연/비해당)
    else:
        return np.nan

df['smoke_grp'] = df.apply(recode_smoking, axis=1)

# 2. 결과 정의: 복부비만 (남성 >= 90cm, 여성 >= 85cm)
def recode_abdominal_obesity(row):
    wc = row['HE_wc']
    sex = row['sex']
    if pd.isna(wc):
        return np.nan
    if sex == 1 and wc >= 90:
        return 1
    elif sex == 2 and wc >= 85:
        return 1
    else:
        return 0

df['abdominal_obesity'] = df.apply(recode_abdominal_obesity, axis=1)

# 3. 음주 공변량 재코딩 (BD1_11)
def recode_alcohol(val):
    if val == 1 or val == 8:
        return 0
    elif val in [2, 3, 4, 5, 6]:
        return 1
    else:
        return np.nan

df['alc_recode'] = df['BD1_11'].apply(recode_alcohol)

# 4. 유산소 신체활동 실천율 (pa_aerobic)
df['pa_recode'] = df['pa_aerobic'].apply(lambda x: x if x in [0, 1] else np.nan)

print("변수 재코딩 완료!")

# %% [markdown]
# ## 2-1. 분석 대상 `in_analysis` (10주차 셀 3). 최종 5,629명 / 현재흡연 765 / 비흡연 4,864이어야 한다

# %% 셀 3
# [셀 3] B-1 대상자 흐름도 인원 계산 및 검증
# 1단계: 전체 참여자
step1 = len(df)

# 2단계: 만 20세 이상 성인
step2_df = df[df['age'] >= 20]
step2 = len(step2_df)
step2_exc = step1 - step2

# 3단계: 노출 정의 가능자 (BS3_1이 1, 2, 3, 8인 사람)
step3_df = step2_df[step2_df['smoke_grp'].notna()]
step3 = len(step3_df)
step2_3_exc = step2 - step3

# 4단계: 최종 분석 대상 (결과 허리둘레 HE_wc까지 있는 사람)
final_df = step3_df[step3_df['HE_wc'].notna()].copy()
final_n = len(final_df)
step3_4_exc = step3 - final_n

# 분석 대상 표시변수(in_analysis) 생성
df['in_analysis'] = 0
df.loc[final_df.index, 'in_analysis'] = 1

# 최종 흡연군 / 비흡연군 인원
n_smoke = (final_df['smoke_grp'] == 1).sum()
n_nonsmoke = (final_df['smoke_grp'] == 0).sum()

print("=== B-1 대상자 흐름도 인원 검증 ===")
print(f"1단계 (전체 참여자): {step1}")
print(f"2단계 (20세 이상 성인): {step2} (제외: {step2_exc})")
print(f"3단계 (노출 정의 가능): {step3} (제외: {step2_3_exc})")
print(f"4단계 (최종 분석 대상): {final_n} (제외: {step3_4_exc})")
print(f" -> 현재흡연군: {n_smoke}, 비흡연군: {n_nonsmoke}")

# %% 셀 4
# 11주차에서 쓰는 변수 이름 (10주차 노트북 변수에 맞춤)
V = dict(y='abdominal_obesity', x='smoke_grp', sex='sex', age='age', incm='incm', edu='edu',
         alc='alc_recode', pa='pa_recode', w='wt_itvex', strata='kstrata', psu='psu', dom='in_analysis')

# %% [markdown]
# ## 3. 완전 사례 표시와 인원 출력

# %% 셀 5
import numpy as np, pandas as pd

# 모형 1·2(보정집합): 성별, 연령, 소득, 교육, 음주  / 모형 3: + 신체활동
set2 = [V['sex'], V['age'], V['incm'], V['edu'], V['alc']]
set3 = set2 + [V['pa']]

df['cc2'] = ((df[V['dom']] == 1) & df[set2 + [V['x'], V['y']]].notna().all(axis=1)).astype(int)
df['cc3'] = ((df[V['dom']] == 1) & df[set3 + [V['x'], V['y']]].notna().all(axis=1)).astype(int)

rows = []
for name, flag in [('Table 1 분석 대상', V['dom']), ('모형 1·2 (완전 사례)', 'cc2'), ('모형 3 (완전 사례)', 'cc3')]:
    d = df[df[flag] == 1]
    rows.append([name, len(d), int((d[V['x']] == 1).sum()), int((d[V['x']] == 0).sum()),
                 d[V['psu']].nunique()])
n_tab = pd.DataFrame(rows, columns=['대상', 'n', '현재흡연', '비흡연', '집락 수'])
n_tab['Table 1 대비 제외'] = n_tab['n'].iloc[0] - n_tab['n']
print(n_tab.to_string(index=False))

# %% [markdown]
# ## 4. 설계 기반 로지스틱 회귀
# 가중치로 계수를 추정하고, 층·집락 선형화(Taylor)로 분산을 구한다. 행은 지우지 않고 도메인 밖은 가중치 0으로 둔다. ※ 10주차 노트북에 설계 기반 SE 함수가 이미 있으면 그것을 쓰고, 아래 함수는 대체용이다. 자유도 = 집락 수 − 층 수.

# %% 셀 6
import statsmodels.api as sm
from scipy import stats

def svy_logit(data, y, xcols, w, strata, psu, domain):
    """domain==1 인 사람만 가중치를 유지(나머지 w=0, 행은 그대로). 반환: 계수표, 도메인 n"""
    d = data.copy()
    d['_w'] = np.where(d[domain] == 1, d[w], 0.0)
    X = pd.get_dummies(d[xcols], columns=[c for c in xcols if d[c].dtype.name in ('category', 'object')],
                       drop_first=True).astype(float)
    X = sm.add_constant(X, has_constant='add')
    inside = d['_w'] > 0
    X = X.where(inside, X.fillna(0))  # 도메인 밖(가중치 0)의 결측이 NaN을 퍼뜨리지 않게 0으로 채움
    yv = d[y].astype(float).fillna(0)

    fit = sm.GLM(yv[inside], X[inside], family=sm.families.Binomial(),
                 freq_weights=d.loc[inside, '_w']).fit()
    beta = fit.params.values
    p = 1 / (1 + np.exp(-(X.values @ beta)))
    wv = d['_w'].values

    # 점수 기여도와 정보행렬 (가중치 0인 행은 자동으로 0)
    U = (wv * (yv.values - p))[:, None] * X.values
    A = (X.values * (wv * p * (1 - p))[:, None]).T @ X.values

    # 층별 집락 합계 -> 층 안에서 중심화 -> n_h/(n_h-1)
    g = pd.DataFrame(U); g['s'] = d[strata].values; g['c'] = d[psu].values
    B = np.zeros_like(A)
    for _, gs in g.groupby('s'):
        tot = gs.drop(columns='s').groupby('c').sum().values
        nh = tot.shape[0]
        if nh > 1:
            z = tot - tot.mean(axis=0)
            B += nh / (nh - 1) * z.T @ z
    Ainv = np.linalg.inv(A)
    V_ = Ainv @ B @ Ainv
    se = np.sqrt(np.diag(V_))

    n_h = d.groupby(strata)[psu].nunique()
    dof = n_h.sum() - n_h.shape[0]
    t = beta / se
    tq = stats.t.ppf(0.975, dof)
    out = pd.DataFrame({'OR': np.exp(beta), 'lo': np.exp(beta - tq * se), 'hi': np.exp(beta + tq * se),
                        'p': 2 * stats.t.sf(np.abs(t), dof)}, index=X.columns)
    return out, int(inside.sum()), int(dof)

# %% 셀 7
def run(xcols, flag):
    res, n, dof = svy_logit(df, V['y'], xcols, V['w'], V['strata'], V['psu'], flag)
    return res, n, dof

# 범주형 공변량은 category로 (소득, 교육)
for c in [V['incm'], V['edu']]:
    df[c] = df[c].astype('category')

m1, n1, dof1 = run([V['x']], 'cc2')            # 모형 1: 노출만, 모형 2와 같은 사람
m2, n2, dof2 = run([V['x']] + set2, 'cc2')     # 모형 2: DAG 보정집합
m3, n3, dof3 = run([V['x']] + set3, 'cc3')     # 모형 3: + 신체활동
m2b, n2b, _  = run([V['x']] + set2, 'cc3')     # 비교용: 모형 3과 같은 사람에서 모형 2
print('자유도(집락-층):', dof1, dof2, dof3)

# %% [markdown]
# ## 5. Table 2 형식으로 정리

# %% 셀 8
def fmt(res, label, n):
    r = res.loc[V['x']]
    return {'모형': label, '분석 n': n, 'OR': f"{r.OR:.2f}", '95% CI': f"{r.lo:.2f}–{r.hi:.2f}", 'p': f"{r.p:.3f}"}

table2 = pd.DataFrame([
    fmt(m1,  '모형 1 (노출만)', n1),
    fmt(m2,  '모형 2 (성별·연령·소득·교육·음주)', n2),
    fmt(m3,  '모형 3 (모형 2 + 신체활동)', n3),
    fmt(m2b, '(참고) 모형 2, 모형 3과 같은 사람', n2b),
])
print(table2.to_string(index=False))
print('\n보정변수 계수는 Table 2에 넣지 않지만 확인용으로 출력:')
print(m2.round(3).to_string())

# %% [markdown]
# ## 6. (참고) 보정변수를 하나씩 더했을 때 OR 변화 — C-3 해석용, Table 2에는 넣지 않음

# %% 셀 9
rows = []
for lab, cols in [('노출만', []), ('+ 성별', [V['sex']]), ('+ 연령', [V['age']]), ('+ 성별·연령', [V['sex'], V['age']]),
                  ('+ 소득·교육', [V['incm'], V['edu']]), ('+ 음주', [V['alc']]), ('+ 성별·연령·소득·교육·음주', set2)]:
    r, n, _ = run([V['x']] + cols, 'cc2')
    rows.append({'보정': lab, 'n': n, 'OR': f"{r.loc[V['x']].OR:.2f}", '95% CI': f"{r.loc[V['x']].lo:.2f}–{r.loc[V['x']].hi:.2f}"})
print(pd.DataFrame(rows).to_string(index=False))
```
