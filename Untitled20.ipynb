```python
import numpy as np
import pandas as pd

# 1. 데이터 로드 및 설계 변수 지정 (knhanes_like_week4.csv)
df = pd.read_csv('knhanes_like_week4.csv')

w = df['wt'].to_numpy()
ST = df['kstrata'].to_numpy()
PS = df['psu'].to_numpy()

# 2. 궁극집락(Ultimate Cluster) 분산 산출 함수
def _var_psu(z, strata, psu):
    t = pd.DataFrame({'st': strata, 'cl': psu, 'z': z})
    zc = t.groupby(['st', 'cl'])['z'].sum().reset_index()
    var = 0.0
    for s, g in zc.groupby('st'):
        m = len(g)
        if m > 1:
            var += m / (m - 1) * np.sum((g['z'] - g['z'].mean()) ** 2)
    return var

# 3. 복합표본 설계를 반영한 도메인 평균/비율 계산 함수
def svymean_domain(y, w, strata, psu, domain):
    y = np.asarray(y, float)
    d = np.asarray(domain, float) * (~np.isnan(y))
    y = np.nan_to_num(y) * d
    wd = np.asarray(w, float) * d
    W_dom = np.sum(wd)
    if W_dom == 0:
        return np.nan, np.nan
    th = np.sum(wd * y) / W_dom
    z = wd * (y - th) / W_dom
    se = np.sqrt(_var_psu(z, strata, psu))
    return th, se

# 4. 도메인 지시변수 생성
ok_drink = df['drink'].notna().to_numpy()
nodrink = ((df['drink'] == 1) & ok_drink).to_numpy()
drinker = ((df['drink'] != 1) & ok_drink).to_numpy()

DOMS = {
    '전체 (Total)': np.ones(len(df), bool),
    '비음주군 (Non-drinker)': nodrink,
    '음주군 (Drinker)': drinker,
}

# 5. 행 생성 함수 (n = 해당 변수의 결측 제외 대상자 수, 범주형도 평균 (SE) 형식)
def row_cont(label, var):
    r = {'변수 (Variables)': label, 'n': int(df[var].notna().sum())}
    for name, dom in DOMS.items():
        th, se = svymean_domain(df[var].to_numpy(), w, ST, PS, dom)
        r[name] = f'{th:.1f} ({se:.2f})'
    return r

def row_cat(label, var, mask):
    # var: 결측 여부를 판단할 원 변수명, mask: 해당 범주 여부(True/False)
    y = np.where(df[var].notna(), np.asarray(mask, float), np.nan)
    r = {'변수 (Variables)': label, 'n': int(df[var].notna().sum())}
    for name, dom in DOMS.items():
        th, se = svymean_domain(y, w, ST, PS, dom)
        r[name] = f'{th * 100:.1f} ({se * 100:.2f})'
    return r

# 6. Table 1 생성
t1_rows = [
    row_cont('연령 (세), 평균 (SE)', 'age'),
    row_cat('65세 이상, % (SE)', 'age', df['age'] >= 65),
    row_cat('남성, % (SE)', 'sex', df['sex'] == 1),
    row_cat('읍면지역 거주, % (SE)', 'urban', df['urban'] == 2),
    row_cont('체질량지수 (kg/m²), 평균 (SE)', 'bmi'),
    row_cat('비만 (BMI ≥ 25), % (SE)', 'bmi', df['bmi'] >= 25),
    row_cont('수축기 혈압 (mmHg), 평균 (SE)', 'sbp'),
    row_cat('고혈압 유병률, % (SE)', 'htn', df['htn'] == 1),
    row_cont('공복혈당 (mg/dL), 평균 (SE)', 'glucose'),
    row_cat('현재흡연, % (SE)', 'smk', df['smk'] == 1),
    row_cat('규칙적 운동, % (SE)', 'exercise', df['exercise'] == 2),
]

t1_df = pd.DataFrame(t1_rows)
print(t1_df.to_markdown(index=False))
```

```python
#비가중 vs 가중 비교표 (전체 대상, 결측 제외)
ALL = np.ones(len(df), bool)

def unweighted_pct(var, mask):
    # 비가중: 표본 안의 단순 비율 (결측 제외)
    ok = df[var].notna().to_numpy()
    return 100 * np.mean(np.asarray(mask, float)[ok])

def weighted_pct(var, mask):
    # 가중: 복합표본 가중치를 적용한 추정치
    y = np.where(df[var].notna(), np.asarray(mask, float), np.nan)
    th, _ = svymean_domain(y, w, ST, PS, ALL)
    return 100 * th

b3_items = [
    ('읍면지역 거주 비율', 'urban', df['urban'] == 2),
    ('65세 이상 고령층 비율', 'age', df['age'] >= 65),
    ('고혈압 유병률', 'htn', df['htn'] == 1),
]

b3_rows = []
for label, var, mask in b3_items:
    u = unweighted_pct(var, mask)
    g = weighted_pct(var, mask)
    b3_rows.append({'변수': label,
                    '비가중 수치 (%)': f'{u:.1f}',
                    '가중 수치 (%)': f'{g:.1f}',
                    '차이 (%p)': f'{g - u:+.1f}'})

print(pd.DataFrame(b3_rows).to_markdown(index=False))
```
