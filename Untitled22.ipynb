```python
!pip install pyreadstat

import pyreadstat
import pandas as pd

# 구글 드라이브에 업로드한 sav 파일 경로
# (아까 MyDrive/hdac_data/ 폴더에 넣었으므로 아래 경로로 지정해 줘)
file_path = '/content/drive/MyDrive/hdac_data/HN24_ALL.sav'

# 변수 정의표에 적어둔 우리 조의 '필요한 변수들만' 골라서 읽어오기 (메모리 절약)
# 예시로 몇 개만 적어둘 테니, 실제 변수명으로 바꿔서 넣어줘!
selected_cols = ['age', 'sex', 'HE_wc', 'BS1_1'] # 예시 변수명 (실제 원변수명으로 수정 필요)

# 데이터 읽기
df, meta = pyreadstat.read_sav(file_path, usecols=selected_cols)

print("데이터 불러오기 성공! 전체 행/열 크기:", df.shape)
```

```python
# 노출 변수(평생흡연여부 BS1_1) 및 결과 변수(허리둘레 HE_wc)의 빈도표 및 기술통계 확인하기
print("--- 평생흡연여부 (BS1_1) 빈도표 ---")
print(df['BS1_1'].value_counts(dropna=False))

print("\n--- 허리둘레 (HE_wc) 기술통계 ---")
print(df['HE_wc'].describe())
```

```python
# 공변량 및 복합표본 변수 확인 예시
print(df['age'].describe())
print(df['sex'].value_counts(dropna=False))
```

```python
# 예시: 미성년자 제외 및 결측치 제거 후 인원 확인
df_filtered = df[(df['age'] >= 19) & (df['HE_wc'].notnull()) & (df['BS1_1'].notnull())]
print("최종 분석 대상자 수:", len(df_filtered))
```
